```python
from PIL import Image, ImageDraw, ImageFont
import pandas as pd
#%pip install pillow 
```

```python
#Archivo Excel
filtro = "online" # "Campo","online", "Guias Mushullakta"

datos = pd.read_excel("datos_certificados.xlsx")
datos = datos[datos["Componente"] == filtro]["Nombre_completo"].to_list() 
#Ahora tenemos que convertilo en una lista.

ci = pd.read_excel("datos_certificados.xlsx")
ci = ci[ci["Componente"] == filtro ]["CI_Mushullakta"].astype(str)#.to_list() #solo para quienes se les pidio los numeros
#ci = ci.replace('.0', '').to_list()
#Cuando ya se tiene un poco más de experiencia podría escribirse todo en una sola línea.
ci = [x.replace('.0', '') for x in ci] 
```

```python
tipo_certificado = "1.jpg" # 1= online, 2 = campo, 3 = Mushullakta certificado campo, online, Mushullakta

coordenadas = (400,600) #600 funciona mejor pero Mushullakta es diferente
coordenadas_ci = (400,650)
#coordenadas (x,y) pueden obtenerse abriendo la imagen previamente en paint.
color_texto = (0,137,209)
#Hay que escribir el código del color en formato (R,G,B) 
tipo_letra = ImageFont.truetype('arial.ttf', 70) 
tipo_letra_ci = ImageFont.truetype('arial.ttf', 45)
#Se especifica el nombre del archivo .tff que guardamos previamente y el tamaño de la letra ("nombre.tff", tamaño)
```

```python
names = range(len(datos))
for i in names:
 certificado = Image.open(tipo_certificado)
 nuevo = ImageDraw.Draw(certificado)
 #nuevo.text(coordenadas_ci, "con cédula de identidad número " + ci[i], fill=color_texto, font=tipo_letra_ci) #activar solo Mushu
 nuevo.text(coordenadas, datos[i], fill=color_texto, font=tipo_letra)
 
 certificado.save(datos[i]+".pdf")
 print("certificado guardado " + datos[i])
```
